```python

# ============================================================
# CNN: HANDWRITTEN DIGIT CLASSIFICATION
# ============================================================

# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import time

from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report
)

# Task 1: Load and explore the dataset

df = pd.read_csv('/content/CNN_Dataset.csv')

X = df.drop(columns=['label']).values.astype('float32')
y = df['label'].values.astype('int32')

# Display class distribution
print("Class distribution:")
print(df['label'].value_counts().sort_index())


# Display sample images before preprocessing
images = X.reshape(-1, 8, 8)

plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(images[i], cmap='gray')
    plt.title(f"Label: {y[i]}")
    plt.axis('off')
plt.tight_layout()
plt.show()

# Inspect image dimensions, colour mode, and pixel range

# Original image dimensions
print("Image dimensions:", (8, 8))

# Colour mode
print("Colour mode: Grayscale")

# Pixel value range
print("Pixel value range:", X.min()," - ", X.max())




# Task 2: Preprocess the images

# Normalize pixel values to [0, 1]
X = X / 16.0

# Resize images from 8x8 to 32x32
X = tf.image.resize(
    X.reshape(-1, 8, 8, 1), (32, 32)
).numpy()

# Display images after preprocessing
plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X[i].squeeze(), cmap='gray', vmin=0, vmax=1)
    plt.title(f"Label: {y[i]}")
    plt.axis('off')
plt.tight_layout()
plt.show()

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

#print("Training images:", len(X_train))
#print("Testing images:", len(X_test))


# Task 3: Design the baseline CNN

def build_cnn():
    model = keras.Sequential([
        keras.layers.Input(shape=(32, 32, 1)),
        keras.layers.Conv2D(16, (3, 3), activation='relu'),
        keras.layers.MaxPooling2D((2, 2)),
        keras.layers.Conv2D(32, (3, 3), activation='relu'),
        keras.layers.MaxPooling2D((2, 2)),
        keras.layers.Flatten(),
        keras.layers.Dense(64, activation='relu'),
        keras.layers.Dense(10, activation='softmax')
    ])
    return model

model = build_cnn()
model.summary()


# Task 4: Compile and train the CNN

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

history = model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.20,
    verbose=1
)

baseline_time = time.time() - start

# Plot training and validation loss
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.show()

# Plot training and validation accuracy
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.show()


# Task 5: Evaluate the trained CNN

probabilities = model.predict(X_test, verbose=0)
y_pred = np.argmax(probabilities, axis=1)

print("Test accuracy:", accuracy_score(y_test, y_pred))
print("Macro precision:", precision_score(
    y_test, y_pred, average='macro', zero_division=0
))
print("Macro recall:", recall_score(
    y_test, y_pred, average='macro', zero_division=0
))
print("Macro F1-score:", f1_score(
    y_test, y_pred, average='macro', zero_division=0
))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

cm = confusion_matrix(y_test, y_pred)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix")
plt.xlabel("Predicted Digit")
plt.ylabel("True Digit")
plt.show()

# Find the most confused digit pair
errors = cm.copy()
np.fill_diagonal(errors, 0)

if errors.max() > 0:
    actual, predicted = np.unravel_index(
        np.argmax(errors), errors.shape
    )
    print(
        "Most confused pair: digit", actual,
        "predicted as", predicted,
        "-", errors[actual, predicted], "times"
    )


# Task 6: Visualize predictions

plt.figure(figsize=(12, 6))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_test[i].squeeze(), cmap='gray')
    color = 'green' if y_test[i] == y_pred[i] else 'red'
    plt.title(
        f"True: {y_test[i]}, Pred: {y_pred[i]}",
        color=color
    )
    plt.axis('off')

plt.tight_layout()
plt.show()

# Display softmax probabilities for selected images
for i in range(3):
    print(f"\nImage {i + 1}")
    print("True label:", y_test[i])
    print("Predicted label:", y_pred[i])
    print("Class probabilities:", np.round(probabilities[i], 3))


# Task 7: Compare with a simpler CNN

simple_model = keras.Sequential([
    keras.layers.Input(shape=(32, 32, 1)),
    keras.layers.Conv2D(16, (3, 3), activation='relu'),
    keras.layers.Flatten(),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

simple_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

simple_model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.20,
    verbose=0
)

simple_time = time.time() - start

simple_pred = np.argmax(
    simple_model.predict(X_test, verbose=0), axis=1
)

simple_accuracy = accuracy_score(y_test, simple_pred)
simple_f1 = f1_score(
    y_test, simple_pred, average='macro', zero_division=0
)


# Task 8: Compare with a dense neural network

X_train_flat = X_train.reshape(len(X_train), -1)
X_test_flat = X_test.reshape(len(X_test), -1)

dense_model = keras.Sequential([
    keras.layers.Input(shape=(1024,)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

dense_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

dense_model.fit(
    X_train_flat, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.20,
    verbose=0
)

dense_time = time.time() - start

dense_pred = np.argmax(
    dense_model.predict(X_test_flat, verbose=0), axis=1
)

dense_accuracy = accuracy_score(y_test, dense_pred)
dense_f1 = f1_score(
    y_test, dense_pred, average='macro', zero_division=0
)


# Compare model performance

comparison = pd.DataFrame({
    'Model': ['Baseline CNN', 'Simple CNN', 'Dense Neural Network'],
    'Test Accuracy': [
        accuracy_score(y_test, y_pred),
        simple_accuracy,
        dense_accuracy
    ],
    'Macro F1-score': [
        f1_score(y_test, y_pred, average='macro', zero_division=0),
        simple_f1,
        dense_f1
    ],
    'Training Time (seconds)': [
        baseline_time, simple_time, dense_time
    ]
})

print("\nModel Comparison:")
print(comparison.round(4).to_string(index=False))


# Task 9: Summarize the results


# Report evaluation metrics
#print(f"Test Accuracy: {accuracy_score(y_test, y_pred):.4f}")
#print(f"Macro Precision: {precision_score(y_test, y_pred, average='macro'):.4f}")
#print(f"Macro Recall: {recall_score(y_test, y_pred, average='macro'):.4f}")
#print(f"Macro F1-score: {f1_score(y_test, y_pred, average='macro'):.4f}")

# Summarize the loss curve
print("\nLoss Curve:")
print("Initial training loss:", round(history.history['loss'][0], 4))
print("Final training loss:", round(history.history['loss'][-1], 4))
print("Initial validation loss:", round(history.history['val_loss'][0], 4))
print("Final validation loss:", round(history.history['val_loss'][-1], 4))

# Summarize the confusion matrix
errors = confusion_matrix(y_test, y_pred).copy()
np.fill_diagonal(errors, 0)

actual, predicted = np.unravel_index(
    np.argmax(errors), errors.shape
)

print("\nConfusion Matrix Finding:")
print(f"Digit {actual} was most often misclassified as {predicted} "
      f"({errors[actual, predicted]} times).")

# Summarize sample predictions
correct = np.sum(y_test == y_pred)
incorrect = np.sum(y_test != y_pred)

print("\nSample Prediction Finding:")
print("Correct predictions:", correct)
print("Incorrect predictions:", incorrect)

#print("\nConclusion:")
#print("The CNN achieved good performance on unseen test images.")
#print("The loss curve shows how the model learned during training.")
#print("The confusion matrix identifies digit classes that need improvement.")

```
